```python
#Гергет Игорь
def split_and_swap_string(input_string):
    middle = len(input_string) // 2

    first_half = input_string[:middle + len(input_string) % 2]
    second_half = input_string[middle + len(input_string) % 2:]

    new_string = second_half + first_half

    return new_string

input_str = "гетГер"

result = split_and_swap_string(input_str)
print(result)
```

```python
#Гергет Игорь
def reverse_between_four(input_string):
    first_index = input_string.index('4')
    last_index = input_string.rindex('4')

    fragment_to_reverse = input_string[first_index + 1:last_index]
    reversed_fragment = fragment_to_reverse[::-1]

    result = input_string[:first_index + 1] + reversed_fragment + input_string[last_index:]
    return result

input_str = "12Игорь4Гергет4"

result = reverse_between_four(input_str)
print(result)
```

```python
#Гергет Игорь
def find_first_and_last_occurrence(input_string):
    char_dict = {char: index for index, char in enumerate(input_string)}

    result = ""
    for char, index in char_dict.items():
        if input_string.count(char) == 1:
            result = f"символ '{char}' встречается один раз, индекс {index}"
        elif input_string.count(char) >= 2:
            first_index = input_string.index(char)
            last_index = input_string.rindex(char)
            result = f"символ '{char}' встречается более одного раза, первое вхождение {first_index}, последнее вхождение {last_index}"

    return result

input_str = "Гергет"

result = find_first_and_last_occurrence(input_str)
print(result)
```

```python
#Гергет Игорь
def check_words():
    words = []
    while True:
        new_word = input("введите слово ")
        words.append(new_word)

        if len(words) > 1:
            if new_word[0] != words[-2][-1]:
                print("последнее корректное слово", words[-2])
                break

check_words()
```

```python
#Гергет Игорь
def word_with_hindrances(word):
    for index, letter in enumerate(word, start=1):
        print(letter * index, end='')

input_word = "Зимин"

word_with_hindrances(input_word)
```

```python
#Гергет Игорь
def stretch_word(word):
    mid = len(word) // 2
    if len(word) % 2 == 0:
        part1 = word[:mid-1]
        part2 = word[mid-1:mid+1]
        part3 = word[mid+1:]
        result = part1 + part2.upper() + part3
    else:
        part1 = word[:mid]
        part2 = word[mid]
        part3 = word[mid+1:]
        result = part1 + part2.upper() + part3
    return result

input_word = "гергет"

stretched_word = stretch_word(input_word)
print(stretched_word)
```

```python
#Гергет Игорь
def find_elements_larger_than_previous(numbers):
    result = [numbers[i] for i in range(1, len(numbers)) if numbers[i] > numbers[i-1]]
    return result

input_numbers = [1, 5, 2, 7, 3, 8, 6]

result = find_elements_larger_than_previous(input_numbers)
print(result)
```

```python
#Гергет Игорь
def find_same_sign_neighbours(numbers):
    for i in range(len(numbers) - 1):
        if numbers[i] * numbers[i+1] > 0:
            print(numbers[i], numbers[i+1])
            break

input_numbers = [-2, 5, -1, -3, 8, 2, 1]

find_same_sign_neighbours(input_numbers)
```

```python
#Гергет Игорь
def swap_neighbours(input_list):
    for i in range(0, len(input_list) - 1, 2):
        input_list[i], input_list[i+1] = input_list[i+1], input_list[i]

    return input_list

my_list = [1, 2, 3, 4, 5, 6, 7]

result_list = swap_neighbours(my_list)

print(result_list)
```

```python
#Гергет Игорь
def find_unique_elements(input_list):
    unique_elements = [x for x in input_list if input_list.count(x) == 1]
    return unique_elements

my_list = [1, 2, 3, 2, 4, 5, 6, 4]

unique_elements = find_unique_elements(my_list)

print(unique_elements)
```

```python
#Гергет Игорь
def create_new_sentence(sentence):
    words = sentence.split()
    capitalized_word = next(word for word in words if word[0].isupper())
    start_index = words.index(capitalized_word) + 1
    new_sentence = ' '.join(words[start_index:])
    new_sentence = new_sentence[0].upper() + new_sentence[1:]

    return new_sentence

original_sentence = "этого не Будет, а вот это будет"

new_sentence = create_new_sentence(original_sentence)

print(new_sentence)
```

```python
#Гергет Игорь
from collections import deque

queue = deque()

while True:
    action = input().split()

    if action[0] == "Кто" and action[1] == "последний?":
        surname = action[-1].replace('"', '').replace('.', '')
        queue.append(surname)
    elif action[0] == "Я" and action[1] == "только" and action[2] == "спросить!":
        surname = action[-1].replace('"', '').replace('.', '')
        queue.appendleft(surname)
    elif action[0] == "Следующий!":
        if queue:
            print("Заходит", queue.popleft() + "!")
        else:
            print("В очереди никого нет.")
```
