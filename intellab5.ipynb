```python
#Гергет Игорь
def factorials(n):
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i
        yield factorial

n = 7
for f in factorials(n):
    print(f)
```

```python
#Гергет Игорь
def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

def square_fibonacci(n):
    fib_gen = fibonacci()
    for _ in range(n):
        fib_num = next(fib_gen)
        yield fib_num ** 2

n = 7
for sf in square_fibonacci(n):
    print(sf)
```

```python
#Гергет Игорь
def arithmetic_operation(operation):
    if operation == '+':
        return lambda x, y: x + y
    elif operation == '-':
        return lambda x, y: x - y
    elif operation == '*':
        return lambda x, y: x * y
    elif operation == '/':
        return lambda x, y: x / y
    else:
        return None

operation = arithmetic_operation('+')
print(operation(1, 4))
```

```python
#Гергет Игорь
def same_by(characteristic, objects):
    if len(objects) == 0:
        return True

    first_value = characteristic(objects[0])
    for obj in objects[1:]:
        if characteristic(obj) != first_value:
            return False

    return True

values = [0, 2, 10, 6]
if same_by(lambda x: x % 2, values):
    print("same")
else:
    print("different")

values = [1, 2, 3, 4]
if same_by(lambda x: x % 2, values):
    print("same")
else:
    print("different")
```

```python
#Гергет Игорь
def print_operation_table(operation, num_rows=9, num_columns=9):
    for row in range(1, num_rows + 1):
        for col in range(1, num_columns + 1):
            result = operation(row, col)
            print(result, end="\t")
        print()

def multiply(row, col):
    return row * col

print_operation_table(multiply, 9, 9)
```

```python
#Гергет Игорь
def ask_password(login, password, success, failure):
    vowels = "aeiouy"
    consonants = "bcdfghjklmnpqrstvwxz"

    login_vowels = [ch for ch in login if ch in vowels]
    login_consonants = [ch for ch in login if ch in consonants]

    password_vowels = [ch for ch in password if ch in vowels]
    password_consonants = [ch for ch in password if ch in consonants]

    if len(password_vowels) != 3 or len(password_consonants) != len(login_consonants):
        if len(password_vowels) != 3 and len(password_consonants) != len(login_consonants):
            failure(login, "Everything is wrong".upper())
        elif len(password_vowels) != 3:
            failure(login, "Wrong number of vowels".upper())
        else:
            failure(login, "Wrong consonants".upper())
    else:
        success(login)


def main(login, password):
    def success(login):
        print(f"Привет, {login.capitalize()}!")

    def failure(login, error_message):
        print(f"Кто-то пытался притвориться пользователем {login.capitalize()}, но в пароле допустил ошибку: {error_message}.")

    ask_password(login.lower(), password.lower(), success, failure)

main("login", "aaalgn")
main("login", "test")
```

```python
#Гергет Игорь
import math


def distance_from_origin(point):
    return math.sqrt(point[0] ** 2 + point[1] ** 2)


def compare_points(p1, p2):
    dist1 = distance_from_origin(p1)
    dist2 = distance_from_origin(p2)

    if dist1 < dist2:
        return -1
    elif dist1 > dist2:
        return 1
    else:
        if p1[0] < p2[0]:
            return -1
        elif p1[0] > p2[0]:
            return 1
        else:
            if p1[1] < p2[1]:
                return -1
            elif p1[1] > p2[1]:
                return 1
            else:
                return 0


points = [(1, 2), (3, 4), (-1, -1), (4, 3)]
sorted_points = sorted(points, key=lambda x: (distance_from_origin(x), x[0], x[1]))

for point in sorted_points:
    print(point)
```

```python
#Гергет Игорь
def check_password(password):
    def decorator(func):
        def wrapper(*args, **kwargs):
            input_password = input("введите пароль ")
            if input_password == password:
                return func(*args, **kwargs)
            else:
                print("доступ запрещен")
                return None
        return wrapper
    return decorator

@check_password('пароль')
def make_burger(typeOfMeat, withOnion=False, withTomato=True):
    print(f"Готовим бургер с {typeOfMeat}, лук: {withOnion}, помидор: {withTomato}")

make_burger("говядина", withOnion=True)
```

```python
#Гергет Игорь
def cached(func):
    cache = {}
    def wrapped(n):
        if n not in cache:
            cache[n] = func(n)
        return cache[n]
    return wrapped

@cached
def fib(n):
    if n == 1 or n == 2:
        return 1
    else:
        return fib(n-1) + fib(n-2)

print(fib(10))
```
