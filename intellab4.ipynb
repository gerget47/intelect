```python
#Гергет Игорь
import math

def distance(x1, y1, x2, y2):
    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

x1, y1, x2, y2 = map(float, input().split())

print(distance(x1, y1, x2, y2))
```

```python
#Гергет Игорь
def numder_to_words(n):
    ones = ['', 'один', 'два', 'три', 'четыре', 'пять', 'шесть', 'семь', 'восемь', 'девять']
    teens = ['', 'одиннадцать', 'двенадцать', 'тринадцать', 'четырнадцать', 'пятнадцать', 'шестнадцать', 'семнадцать', 'восемнадцать', 'девятнадцать']
    tens = ['', '', 'двадцать', 'тридцать', 'сорок', 'пятьдесят', 'шестьдесят', 'семьдесят', 'восемьдесят', 'девяносто']
    if n < 10:
        return ones[n]
    elif n < 20:
        return teens[n % 10]
    else:
        return tens[n // 10] + ' ' + ones[n % 10]

print(numder_to_words(15))
print(numder_to_words(19))
print(numder_to_words(23))
print(numder_to_words(34))
print(numder_to_words(45))
print(numder_to_words(56))
print(numder_to_words(78))
print(numder_to_words(89))
```

```python
#Гергет Игорь
def bracket_check(test_string):
    stack = []
    for char in test_string:
        if char == '(':
            stack.append(char)
        elif char == ')':
            if not stack:
                return 'NO'
            stack.pop()
    if stack:
        return 'NO'
    return 'YES'

print(bracket_check("()"))
print(bracket_check("(()))"))
print(bracket_check("(((()"))
print(bracket_check(")("))
print(bracket_check(""))
```

```python
#Гергет Игорь
def palindrome(s):
    s = s.lower()
    s = ''.join(c for c in s if c.isalnum())
    return 'палиндром' if s == s[::-1] else 'не палиндром'

print(palindrome("А роза упала на лапу Азора"))
print(palindrome("1232 1"))
print(palindrome("12345"))
print(palindrome("А роза упала на лапу Азораа"))
```

```python
#Гергет Игорь
def tic_tac_toe(field):
    for row in field:
        if row.count(row[0]) == 3 and row[0] != '*':
            return f"{row[0]} win"

    for col in range(3):
        if field[0][col] == field[1][col] == field[2][col] != '*':
            return f"{field[0][col]} win"

    if field[0][0] == field[1][1] == field[2][2] != '*':
        return f"{field[0][0]} win"
    if field[0][2] == field[1][1] == field[2][0] != '*':
        return f"{field[2][0]} win"

    return "draw"

game_field = [
    ['x', 'x', 'x'],
    ['0', '*', '0'],
    ['*', '0', '*']
]

result = tic_tac_toe(game_field)
print(result)
```

```python
#Гергет Игорь
def print_without_duplicates(message):
    if not hasattr(print_without_duplicates, 'prevmessage'):
        print(message)
        print_without_duplicates.prevmessage = message
    elif print_without_duplicates.prevmessage != message:
        print(message)
        print_without_duplicates.prevmessage = message

print_without_duplicates("Привет")
print_without_duplicates("Не могу до тебя дозвониться")
print_without_duplicates("Не могу до тебя дозвониться")
print_without_duplicates("Не могу до тебя дозвониться")
print_without_duplicates("Когда доедешь до дома")
print_without_duplicates("Ага, жду")
print_without_duplicates("Ага, жду")
```

```python
#Гергет Игорь
friends = {}

def add_friends(name_of_person, list_of_friends):
    friends[name_of_person] = list_of_friends

def are_friends(name_of_person1, name_of_person2):
    if name_of_person2 in friends.get(name_of_person1, []):
        return True
    return False

def print_friends(name_of_person):
    if name_of_person in friends:
        print(' '.join(sorted(friends[name_of_person])))

add_friends("Игорь", ["Александр", "Павел"])
print(are_friends("Игорь", "Антон"))
add_friends("Игорь", ["Антон"])
print(are_friends("Игорь", "Антон"))
```

```python
#Гергет Игорь
def roman():
    global three
    three = one + two

    roman_num = {1: 'I', 4: 'IV', 5: 'V', 9: 'IX', 10: 'X', 40: 'XL', 50: 'L', 90: 'XC', 100: 'C',
                 400: 'CD', 500: 'D', 900: 'CM', 1000: 'M'}

    def int_to_roman(n):
        result = ''
        for value in sorted(roman_num.keys(), reverse=True):
            while n >= value:
                result += roman_num[value]
                n -= value
        return result

    roman_one = int_to_roman(one)
    roman_two = int_to_roman(two)
    roman_three = int_to_roman(three)

    print(f"{roman_one}+{roman_two}={roman_three}")

one = 5
two = 4
roman()
```

```python
#Гергет Игорь
def continue_fibonacci_sequence(sequence, n):
    for i in range(n):
        next_element = sequence[-1] + sequence[-2]
        sequence.append(next_element)
    return sequence

sequence1 = [1, 1]
result1 = continue_fibonacci_sequence(sequence1, 1)
print(*result1)

sequence2 = [1, 1, 2, 3, 5]
result2 = continue_fibonacci_sequence(sequence2, 0)
print(*result2)
```

```python
#Гергет Игорь
def mirror(arr):
    mirrored_part = arr[::-1]
    arr += mirrored_part
    return arr

arr1 = [1, 2]
result1 = mirror(arr1)
print(*result1)

arr2 = [1]
result2 = mirror(arr2)
print(*result2)
```

```python
#Гергет Игорь
def from_string_to_list(string, container):
    numbers = string.split()
    for num in numbers:
        try:
            container.append(int(num))
        except ValueError:
            pass

a1 = [1, 2, 3]
from_string_to_list("1 3 99 52", a1)
print(*a1)

a2 = [77, 'abc']
from_string_to_list("", a2)
print(*a2)
```

```python
#Гергет Игорь
def transpose(matrix):
    transposed = list(zip(*matrix))

    matrix.clear()
    matrix.extend([list(row) for row in transposed])

matrix1 = [[1]]
transpose(matrix1)
for line in matrix1:
    print(*line)

matrix2 = [[1, 2], [3, 4]]
transpose(matrix2)
for line in matrix2:
    print(*line)
```

```python
#Гергет Игорь
def swap(first, second):
    first[:], second[:] = second[:], first[:]

first = [1, 2, 3]
second = [4, 5, 6]
first_content = first[:]
second_content = second[:]
swap(first, second)
print(first, second_content, first == second_content)
print(second, first_content, second == first_content)

first = [1, 2, 3]
second = [4, 5, 6, 7]
first_content = first[:]
second_content = second[:]
swap(first, second)
print(first, second_content, first == second_content)
print(second, first_content, second == first_content)
```

```python
#Гергет Игорь
def fractal_print(obj):
    result = []
    for item in obj:
        if isinstance(item, list) and item is obj:
            result.append("[...]")
        elif isinstance(item, list):
            sub_result = []
            for sub_item in item:
                if isinstance(sub_item, list) and sub_item is item:
                    sub_result.append("[...]")
                else:
                    sub_result.append(sub_item)
            result.append(sub_result)
        else:
            result.append(item)
    print(result)

fractal1 = [3]
fractal1.append(fractal1)
fractal_print(fractal1)

fractal2 = [3]
fractal2.append(fractal2)
fractal2.append(2)
fractal_print(fractal2)
```

```python
#Гергет Игорь
def matrix(n=1, m=None, a=0):
    if m is None:
        m = n

    mat = [[a for _ in range(m)] for _ in range(n)]

    return mat

rows1 = matrix()
for row in rows1:
    print(*row)

rows2 = matrix(2)
for row in rows2:
    print(*row)
```

```python
#Гергет Игорь
def partial_sums(*args):
    result = [0]
    current_sum = 0

    for num in args:
        current_sum += num
        result.append(current_sum)

    return result

print(partial_sums(13))

print(partial_sums(1, 0.5, 0.25, 0.125))
```

```python
#Гергет Игорь
def solve(*coefficients):
    if len(coefficients) == 3:
        a, b, c = coefficients
        if a == 0:
            return [-c / b] if b != 0 else ["all"]
        discriminant = b**2 - 4*a*c
        if discriminant > 0:
            return sorted([(-b + discriminant**0.5) / (2*a), (-b - discriminant**0.5) / (2*a)])
        elif discriminant == 0:
            return [-b / (2*a)]
        else:
            return []
    elif len(coefficients) == 2:
        b, c = coefficients
        if b == 0:
            return ["all"] if c == 0 else []
        else:
            return [-c / b]
    elif len(coefficients) == 1:
        c = coefficients[0]
        return ["all"] if c == 0 else []
    else:
        return None

print(sorted(solve(1, 2, 1)))  # [-1.0]
print(sorted(solve(1, -3, 2)))  # [1.0, 2.0]
```
